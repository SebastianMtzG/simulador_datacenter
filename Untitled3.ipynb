```python
import pandas as pd

df = pd.read_csv("final_dataset_std.csv", sep=";")

print("Filas y columnas:", df.shape)
print("\nColumnas:")
print(df.columns.tolist())
```

```python
df_pequeno = df.iloc[:500].copy()

print("Tamaño del dataset pequeño:", df_pequeno.shape)
```

```python
# Variables que utilizaremos como entradas
columnas_X = [
    c for c in df_pequeno.columns
    if c != "TLHC"
]

# X contiene las entradas
X = df_pequeno[columnas_X].values

# y contiene la temperatura que queremos predecir
y = df_pequeno["TLHC"].values

print("Forma de X:", X.shape)
print("Forma de y:", y.shape)
```

```python
print(df_pequeno.columns.tolist())
```

```python
import numpy as np

variables = ["P_ac", "P_cu", "T_out", "T_MEAS", "T_celCC"]

X_lista = []

for _, fila in df_pequeno.iterrows():
    secuencia = []

    for t in range(8):
        paso = [
            fila[f"{variable}-{t}"]
            for variable in variables
        ]
        secuencia.append(paso)

    X_lista.append(secuencia)

X = np.array(X_lista, dtype=np.float32)
y = df_pequeno["TLHC"].to_numpy(dtype=np.float32)

print("Forma de X:", X.shape)
print("Forma de y:", y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense

modelo = Sequential([
    Input(shape=(8, 5)),
    LSTM(16),
    Dense(1)
])

modelo.summary()
```

```python
modelo.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
historial = modelo.fit(
    X,
    y,
    epochs=30,
    batch_size=16,
    verbose=1
)
```
