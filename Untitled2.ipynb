```python
import numpy as np
import matplotlib.pyplot as plt

temperaturas = np.array([
    20, 21, 22, 23, 24,
    23, 22, 21, 20, 21,
    22, 23, 24, 25, 24
])

plt.plot(temperaturas)
plt.xlabel("Tiempo")
plt.ylabel("Temperatura °C")
plt.show()
```

```python
ventana = 4

X = []
y = []

for i in range(len(temperaturas) - ventana):
    X.append(temperaturas[i:i+ventana])
    y.append(temperaturas[i+ventana])

X = np.array(X)
y = np.array(y)

print("X:")
print(X)

print("\ny:")
print(y)
```
