```python
import numpy as np
import matplotlib.pyplot as plt

# Secuencia
T = 30
x = np.ones(T)

# Peso recurrente pequeño
W = 0.5

# Forward
h = 1.0
estados = []

for t in range(T):
    h = np.tanh(W * h + x[t])
    estados.append(h)

# Backpropagation simplificado
gradiente = 1.0
gradientes = []

for t in range(T - 1, -1, -1):

    h_t = estados[t]

    derivada_tanh = 1 - h_t**2

    gradiente = gradiente * W * derivada_tanh

    gradientes.append(abs(gradiente))

gradientes = gradientes[::-1]

print("Gradientes:")
print(gradientes)

plt.plot(gradientes)
plt.yscale("log")
plt.xlabel("Paso temporal")
plt.ylabel("Magnitud del gradiente")
plt.title("Gradiente durante BPTT")
plt.grid()
plt.show()
```
